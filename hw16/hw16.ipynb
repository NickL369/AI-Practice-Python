```python
nlp = spacy.load("uk_core_news_lg")

text = "Для приготування Біг Маку в Макдональдз використовують дві котлети з яловичини, спеціальний соус, цибулю, салат айсберг, сир Чеддер та мариновані огірки. Все це подається на тришаровій булочці з кунжутом. Рей Крок зробив цей рецепт стандартом у США, а тепер його готують і в Україні."

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)

displacy.render(doc, style="ent", jupyter=True)
```
