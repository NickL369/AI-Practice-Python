```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_class_weight

df = pd.read_csv("data/glass.data", header=None, names=["id","RI","Na","Mg","Al","Si","K","Ca","Ba","Fe","Type"])

X = df.drop(columns=["id","Type"])
y = df["Type"].astype(int)

X["Na_Mg_ratio"] = X["Na"] / (X["Mg"] + 1e-6)
X["Ca_Si_ratio"] = X["Ca"] / (X["Si"] + 1e-6)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model_plain = LogisticRegression(max_iter=2000, n_jobs=1)
model_plain.fit(X_train, y_train)
pred_plain = model_plain.predict(X_test)

print("LogReg without balancing")
print(classification_report(y_test, pred_plain))
print("Accuracy:", accuracy_score(y_test, pred_plain))

classes = np.unique(y_train)
weights = compute_class_weight(class_weight="balanced", classes=classes, y=y_train)
class_weights = dict(zip(classes, weights))

model_weighted = LogisticRegression(max_iter=2000, class_weight=class_weights, n_jobs=1)
model_weighted.fit(X_train, y_train)
pred_weighted = model_weighted.predict(X_test)

print("\nLogReg with class_weight=balanced")
print(classification_report(y_test, pred_weighted))
print("Accuracy:", accuracy_score(y_test, pred_weighted))

rf_plain = RandomForestClassifier(n_estimators=50, max_depth=8, random_state=42, n_jobs=1)
rf_plain.fit(X_train, y_train)
rf_plain_pred = rf_plain.predict(X_test)

print("\nRandomForest without balancing")
print(classification_report(y_test, rf_plain_pred))
print("Accuracy:", accuracy_score(y_test, rf_plain_pred))

rf_balanced = RandomForestClassifier(n_estimators=50, max_depth=8, class_weight="balanced", random_state=42, n_jobs=1)
rf_balanced.fit(X_train, y_train)
rf_balanced_pred = rf_balanced.predict(X_test)

print("\nRandomForest with balancing")
print(classification_report(y_test, rf_balanced_pred))
print("Accuracy:", accuracy_score(y_test, rf_balanced_pred))
```
